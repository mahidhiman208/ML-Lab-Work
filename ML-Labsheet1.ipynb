```python
# Program 1: Print statement
print("Program 1: Welcome to Machine Learning Lab")

# Program 2: Variable Assignment
x = 10
y = 20
print("\nProgram 2 - Sum:", x + y)

# Program 3: Data Types
a, b, c = 5, 3.14, "ML"
print(f"\nProgram 3 - Types: {type(a)}, {type(b)}, {type(c)}")

# Program 4: User Input Simulation
age = 22
print("\nProgram 4 - User Age:", age)

# Program 5: If-Else Condition
if age >= 18:
    print("\nProgram 5: Eligible for Voting")
else:
    print("\nProgram 5: Not Eligible")

# Program 6: For Loop
print("\nProgram 6 - Loop 1 to 5:")
for i in range(1, 6):
    print(i, end=" ")
print()

# Program 7: While Loop
count = 3
print("\nProgram 7 - Countdown:")
while count > 0:
    print(count, end=" ")
    count -= 1
print()

# Program 8: Function Definition
def greet(name):
    return f"Hello, {name}!"
print("\nProgram 8:", greet("Student"))

# Program 9: List Operations
lst = [10, 20, 30]
lst.append(40)
print("\nProgram 9 - Updated List:", lst)

# Program 10: Dictionary Usage
student = {"name": "Aman", "marks": 85}
print("\nProgram 10 - Student Name:", student["name"])
```

```python
import numpy as np

# Program 11: NumPy Array Creation
arr = np.array([1, 2, 3, 4, 5])
print("Program 11 - NumPy Array:", arr)

# Program 12: Array Shape and Dimension
print("\nProgram 12 - Shape:", arr.shape, "| Dimensions:", arr.ndim)

# Program 13: Array Reshaping
arr_2d = np.arange(6).reshape(2, 3)
print("\nProgram 13 - Reshaped 2D Array:\n", arr_2d)

# Program 14: Element-wise Arithmetic
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print("\nProgram 14 - Addition:", a + b)

# Program 15: Statistical Measures
data = np.array([10, 20, 30, 40, 50])
print(f"\nProgram 15 - Mean: {np.mean(data)}, Std: {np.std(data)}")

# Program 16: Indexing and Slicing
print("\nProgram 16 - Slice (1 to 3):", data[1:4])

# Program 17: Matrix Multiplication
m1 = np.array([[1, 2], [3, 4]])
m2 = np.array([[5, 6], [7, 8]])
print("\nProgram 17 - Matrix Product:\n", np.dot(m1, m2))

# Program 18: Zeroes and Ones Array
print("\nProgram 18 - Zeros Array:\n", np.zeros((2, 2)))

# Program 19: Random Number Generation
np.random.seed(42)
print("\nProgram 19 - Random Numbers:\n", np.random.rand(3))

# Program 20: Finding Min, Max, and Sum
print(f"\nProgram 20 - Min: {data.min()}, Max: {data.max()}, Sum: {data.sum()}")
```

```python
import pandas as pd

# Program 21: Creating DataFrame from Dict
data_dict = {'Name': ['Aman', 'Priya', 'Rohan', 'Neha'], 'Age': [21, 22, 20, 24], 'Marks': [85, 92, 78, 88]}
df = pd.DataFrame(data_dict)
print("Program 21 - DataFrame Created")

# Program 22: Viewing Top Rows
print("\nProgram 22 - Head:\n", df.head(2))

# Program 23: DataFrame Info and Summary
print("\nProgram 23 - Summary Stats:\n", df.describe())

# Program 24: Selecting Columns
print("\nProgram 24 - Marks Column:\n", df['Marks'])

# Program 25: Filtering Data
print("\nProgram 25 - Age > 21:\n", df[df['Age'] > 21])

# Program 26: Adding New Column
df['Passed'] = df['Marks'] >= 80
print("\nProgram 26 - Column Added:\n", df)

# Program 27: Handling Missing Values
df_missing = df.copy()
df_missing.loc[1, 'Marks'] = np.nan
print("\nProgram 27 - Filled Missing Values:\n", df_missing.fillna(0))

# Program 28: GroupBy Operations
print("\nProgram 28 - GroupBy Passed Mean:\n", df.groupby('Passed')['Marks'].mean())

# Program 29: Sorting Values
print("\nProgram 29 - Sorted by Marks:\n", df.sort_values(by='Marks', ascending=False))

# Program 30: Exporting to CSV
df.to_csv("students.csv", index=False)
print("\nProgram 30: Saved to students.csv successfully.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# Program 31: Read CSV File
df_csv = pd.read_csv("students.csv")
print("Program 31 - Loaded CSV:\n", df_csv.head(2))

# Program 32: Scikit-learn Dataset
iris = load_iris()
iris_df = pd.DataFrame(iris.data, columns=iris.feature_names)
print("\nProgram 32 - Iris Dataset Head:\n", iris_df.head(2))

# Program 33: Line Plot
plt.figure(figsize=(5, 3))
plt.plot(df['Age'], df['Marks'], marker='o', color='b')
plt.title("Program 33: Line Plot (Age vs Marks)")
plt.xlabel("Age")
plt.ylabel("Marks")
plt.grid(True)
plt.show()

# Program 34: Histogram
plt.figure(figsize=(5, 3))
plt.hist(df['Marks'], bins=4, color='orange', edgecolor='black')
plt.title("Program 34: Marks Distribution Histogram")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()

# Program 35: Correlation Matrix & Heatmap
plt.figure(figsize=(5, 3))
sns.heatmap(df[['Age', 'Marks']].corr(), annot=True, cmap='coolwarm')
plt.title("Program 35: Correlation Heatmap")
plt.show()

print("\n--- All 35 Programs Executed Cleanly! ---")
```
